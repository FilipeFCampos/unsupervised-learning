## Predição dos Embeddings

Utilize o modelo BERT pré-treinado para gerar embeddings de todos os textos fornecidos.  
O objetivo é obter uma matriz `X` com formato **(N, dim)**, onde **N** é o número de textos e **dim** é a dimensionalidade dos vetores de embedding.

```python
# Importar bibliotecas
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
import numpy as np

import torch
from sentence_transformers import SentenceTransformer

# Utilizar GPU se disponível
device = 'cuda' if torch.cuda.is_available() else 'cpu'

# Carregar o modelo pré-treinado
model = SentenceTransformer('sentence-transformers/all-MiniLM-L6-v2', device)

print(f"Dimensão do embedding: {model.get_sentence_embedding_dimension()}")
```

```python
# Gera os embeddings para as frases
embeddings = model.encode(sentences)

print(f"Shape dos embeddings: {embeddings.shape}")
```

## PCA

Aplique **PCA (Principal Component Analysis)** para projetar os embeddings em duas dimensões e visualizar a estrutura global dos dados.  
O PCA ajuda a capturar as direções de maior variância e pode indicar agrupamentos lineares.

**Tarefas:**
- Reduza a dimensionalidade dos embeddings para 2 componentes principais.  
- Plote os pontos resultantes com `matplotlib`, identificando possíveis agrupamentos.  
- Analise qualitativamente se há separação entre textos de temas distintos.

```python
from sklearn.decomposition import PCA

model_pca = PCA(n_components=2, random_state=42)
embeddings_pca = model_pca.fit_transform(embeddings)

print(f"Shape dos embeddings após o PCA: {embeddings_pca.shape}")
```

```python
def plot_embeddings_2d(embeddings_2d, sentences, title="Projeção dos Embeddings", cmap=None):
    plt.figure(figsize=(10, 10))
    plt.scatter(embeddings_2d[:, 0], embeddings_2d[:, 1], s=100, alpha=0.5, cmap=cmap)

    # Desenha o texto de cada ponto
    for (x, y), text in zip(embeddings_2d, sentences):
        plt.text(x, y, text, fontsize=8)

    plt.title(title)
    plt.xlabel("Componente principal 1")
    plt.ylabel("Componente principal 2")
    plt.grid(True)
    plt.show()
```

```python
plot_embeddings_2d(embeddings_pca, sentences, title="Projeção dos Embeddings PCA")
```

## t-SNE

Use **t-SNE (t-distributed Stochastic Neighbor Embedding)** para investigar a estrutura local dos dados.  
Diferente do PCA, o t-SNE tenta preservar vizinhanças locais e pode revelar grupos mais sutis.

**Tarefas:**
- Reduza os embeddings para 2D usando `TSNE` do `scikit-learn`.  
- Ajuste parâmetros como `perplexity` e `learning_rate` para comparar resultados.  
- Visualize o mapa e observe se os textos semelhantes ficam próximos.

```python
from sklearn.manifold import TSNE

model_tsne = TSNE(n_components=2, perplexity=30, random_state=42, init='random', learning_rate=200)
embeddings_tsne = model_tsne.fit_transform(embeddings)

print(f"Shape dos embeddings após o TSNE: {embeddings_tsne.shape}")
```

```python
plot_embeddings_2d(embeddings_tsne, sentences, title="Projeção dos Embeddings TSNE")
```

## UMAP

Aplique **UMAP (Uniform Manifold Approximation and Projection)** como alternativa ao t-SNE.  
O UMAP é mais eficiente, preserva parte da estrutura global e é útil para visualização e pré-processamento.

**Tarefas:**
- Gere uma projeção 2D dos embeddings com `umap.UMAP`.  
- Experimente variar `n_neighbors` e `min_dist` para observar mudanças na distribuição dos clusters.  
- Compare visualmente com os resultados do PCA e t-SNE.

```python
import umap.umap_ as umap

model_umap = umap.UMAP(n_neighbors=30, min_dist=0.01 ,n_components=2, n_epochs=500, random_state=42)
embeddings_umap = model_umap.fit_transform(embeddings)

print(f"Shape dos embeddings após o UMAP: {embeddings_umap.shape}")
```

```python
plot_embeddings_2d(embeddings_umap, sentences, title="Projeção dos Embeddings UMAP")
```

## Classificação

Com base nas categorias observadas nos gráficos anteriores, crie uma função simples que receba um texto e classifique-o na categoria mais provável.

**Tarefas:**
- Use os embeddings existentes e os clusters identificados para rotular automaticamente cada texto.  
- Crie uma função `classificar_texto(texto: str)` que:
  1. Gere o embedding do texto.
  2. Calcule a distância para os clusters identificados.
  3. Retorne o nome do cluster mais próximo.

O UMAP foi o algoritmo que melhor separou os clusters, então utilizarei a projeção dele para a classificação por meio do kmeans

```python
from sklearn.cluster import KMeans

model_kmeans = KMeans(n_clusters=4, max_iter=150, random_state=42)
model_kmeans.fit(embeddings_umap)
```

```python
def plot_clusters(clusters, embeddings_2d, sentences, n_clusters=4, title="Clusters"):
    plt.figure(figsize=(10, 10))

    sns.scatterplot(
        x=embeddings_2d[:, 0],
        y=embeddings_2d[:, 1],
        hue=clusters,
        palette='viridis',
        s=120
    )

    for (x, y), text in zip(embeddings_2d, sentences):
        plt.text(x, y, text, fontsize=7, alpha=0.75)

    plt.title(f"{title} ({n_clusters} clusters)")
    plt.xlabel("Componente principal 1")
    plt.ylabel("Componente principal 2")
    plt.grid(True)
    plt.legend(title="Cluster")
    plt.show()
```

```python
clusters = model_kmeans.predict(embeddings_umap)

plot_clusters(clusters, embeddings_umap, sentences, title="UMAP + KMEANS")
```

```python
def classificar_texto(text: str):
  high_dim_text = model.encode([text], show_progress_bar=True)
  text_2d = model_umap.transform(high_dim_text)

  prediction = model_kmeans.predict(text_2d)[0]

  return prediction
```

```python
text = 'The stock market is about to crash'

cluster = classificar_texto(text)
print(f"A frase: '{text}' foi classificada no Cluster: {cluster}")
```
